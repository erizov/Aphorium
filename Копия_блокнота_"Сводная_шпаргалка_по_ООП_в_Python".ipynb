## **1.** Конструктор

* **Класс (Class):** Чертеж объекта.
* **Объект (Instance):** Экземпляр, созданный по чертежу.
* **`self`:** Ссылка на текущий объект (экземпляр).
* **`__init__`:** Метод-конструктор для инициализации атрибутов.

```
class Car:
    def __init__(self, brand, color):
        self.brand = brand  # Атрибут экземпляра
        self.color = color

    def drive(self):       # Метод экземпляра
        return f"{self.brand} поехала!"

my_car = Car("Tesla", "Red")
```

### 2.1. Инкапсуляция (Скрытие данных)

* `_name`: **Protected** (защищенный). Соглашение: «не трогай вне класса».
* `__name`: **Private** (приватный). Python искажает имя, чтобы закрыть прямой доступ.


### 2.2. Наследование

Создание класса на базе другого для повторного использования кода.

```
class ElectricCar(Car): # Наследование от Car
    def charge(self):
        return "Заряжаюсь..."
```

### 2.3. Полиморфизм

Разные классы могут иметь методы с одинаковыми именами, но разной логикой.

```
animals = [Dog(), Cat()]
for a in animals:
    a.make_sound() # Каждый издаст свой звук
```

### 2.4. Абстракция (ABC)

Создание «чертежей для чертежей» через модуль `abc`. Обязывает наследников реализовать нужные методы.

```
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self): pass

class Square(Shape):
    def area(self): return 10 * 10 # Обязательная реализация
```

Важно: `isinstance` учитывает наследование, в отличие от `type()`.

## **4.** Множественное наследование и MRO

Python позволяет наследоваться от нескольких классов. Порядок поиска методов определяется алгоритмом **MRO** (Method Resolution Order).

* **`super()`**: Вызывает метод родительского класса (учитывая MRO).
* **`Class.mro()`**: Показывает цепочку поиска методов.


```
class A:
    @classmethod
    def who(cls):
        return "A"

class B(A):
    @classmethod
    def who(cls):
        return "B" + super().who()

class C(A):
    @classmethod
    def who(cls):
        return "C" + super().who()

class D(B, C):
    @classmethod
    def who(cls):
        return "D" + super().who()

    @classmethod
    def other(cls):
        return "D" + super(B, cls).who() # Инструкция super(B, cls) говорит: «Возьми список MRO и начни искать метод who() в классах, которые идут строго после B»

print(D.who())
print(D.other())

print(D.__mro__)
print(D.mro())
```

## **5.** Декораторы: `@property`, `@classmethod`, `@staticmethod`

* **`@property`**: Делает метод доступным как атрибут (геттер). Позволяет добавить логику (валидацию) при чтении/записи.
* **`@classmethod`**: Принимает класс (`cls`). Используется для "фабричных методов" (создание объекта из строки, файла и т.д.).
* **`@staticmethod`**: Обычная функция, которая логически связана с классом, но не нуждается в доступе к `self` или `cls`.

### Основные различия

| Параметр              | `self`                                            | `cls`                                                                     |
|:----------------------|:--------------------------------------------------|:--------------------------------------------------------------------------|
| **Что это?**          | Ссылка на экземпляр (объект) класса.              | Ссылка на сам класс.                                                      |
| **Где используется?** | В обычных методах (экземпляра).                   | В методах класса (@classmethod).                                          |
| **Доступ к данным**   | Видит данные конкретного объекта и данные класса. | Видит только данные класса, не знает о существовании конкретных объектов. |


### Когда что использовать?

* Нужны данные конкретного объекта? — **`self`**.
* Нужно изменить состояние всего класса или создать новый экземпляр? — **`@classmethod` / `cls**`.
* Нужна логика, связанная с темой класса, но не трогающая его данные? — **`@staticmethod`**.
* Нужно вычислить значение и выдать его как переменную? — **`@property`**.

```
class Machine:
    brand = "Generic"

    def __init__(self, model, speed):
        self.model = model
        self._speed = speed  # Нижнее подчеркивание = "приватный" атрибут

    # Геттер (Чтение)
    @property
    def speed(self):
        return self._speed

    # Сеттер (Запись с проверкой)
    @speed.setter
    def speed(self, value):
        if value < 0:
            print("Ошибка: Скорость не может быть отрицательной!")
        else:
            self._speed = value

    # Статический метод
    @staticmethod
    def is_safe(speed):
        return speed <= 120

    # Метод класса (cls)
    @classmethod
    def fast_version(cls, model):
        return cls(model, speed=300)

    # Добавляем **kwargs, чтобы метод принимал любые доп. аргументы
    @classmethod
    def create_default(cls, model, **kwargs):
        # Передаем model, стандартную скорость и всё, что прилетело в kwargs
        return cls(model=model, speed=100, **kwargs)

########## НАСЛЕДОВАНИЕ ##########

class ElectricCar(Machine):
    brand = "Tesla" # Переопределяем атрибут класса

    def __init__(self, model, speed, battery_cap):
        # super() прокидывает self в родительский __init__
        super().__init__(model, speed)
        self.battery = battery_cap

    # 2. Переопределение метода (Polymorphism)
    @property
    def speed(self):
        # Вызываем логику родителя через super() и добавляем свое
        return f"{super().speed} км/ч (Электро)"

########## Проверка ##########

car = Machine("Model S", 100)

# Читаем через @property (без скобок)
print("1.", car.speed)  # Вывод: 100 км/ч

# Пытаемся записать через @speed.setter
print("2.")
car.speed = 150   # Работает!
print("3.", end=" ")
car.speed = -20   # Выведет: Ошибка: Скорость не может быть отрицательной!

# Создаем через @classmethod
print("4.", end=" ")
sport_car = Machine.fast_version("Roadster")
print(sport_car.speed) # Вывод: 300 км/ч

# --- ПРОВЕРКА ---

# Используем @classmethod родителя для создания дочернего объекта! Передаем недостающий аргумент через метод класса
tess = ElectricCar.create_default("Model 3", battery_cap=75)
tess.battery = 75 # Добавляем специфичный атрибут

print("5.", f"Бренд: {tess.brand}")   # Выведет: Tesla
print("6.", f"Скорость: {tess.speed}") # Выведет: 100 км/ч (Электро)
```

## **7.** Магические (Dunder) методы

Позволяют объектам взаимодействовать со встроенными функциями Python.

* `__str__`: (string) Для `print(obj)` и `str(obj)` - Как этот объект выглядит для человека?
* `__repr__`: (representation) Должен показывать, как создать объект (в идеале — быть валидным кодом Python) - Как этот объект выглядит для Python?
| Если определен... | Вызов `str()` или `print()` | Вызов `repr()` или вывод в консоли |
| --- | --- | --- |

| **Оба метода** | Использует `__str__` | Использует `__repr__` |

## **9**. Приоритеты атрибутов в Python


### 9.1. \_\_dict\_\_, \_\_slots\_\_, и Протокол дескрипторов (\_\_get\_\_, \_\_set\_\_).

* `__disc__`: Обычный словарь, который есть у каждого объекта по умолчанию.
* `__slots__`: Это способ явно объявить атрибуты экземпляра, запрещая создание `__dict__` и экономя память

### 9.2. Разница механизмов хранения

| Инструмент | Место | Суть | Память |
| --- | --- | --- | --- |
| `__dict__` | Экземпляр | Динамический словарь. Можно добавлять что угодно. | Высокое потребление |
| `__slots__` | Класс | Статический массив имен. Запрещает `__dict__`. | **Низкое (оптимизация)** |


### 9.3. Кто главный? (*Lookup Chain*)
Когда вы вызываете `obj.attribute`, Python следует строгому алгоритму поиска.<br/>
Если на одном уровне имя найдено, поиск прекращается.

**Иерархия приоритетов (от высшего к низшему):**

Data Descriptor (Дескриптор данных): Класс, у которого есть и `__get__`, и `__set__`.<br/>
`__slots__`: Если атрибут зарезервирован в слотах.<br/>
`__dict__` экземпляра: Обычные атрибуты объекта.<br/>
Non-Data Descriptor: Объект только с `__get__` (например, обычный метод или `@staticmethod`).<br/>
Атрибуты класса: Простые переменные, объявленные в теле класса.<br/>
`__getattr__`: "Запасной выход", если ничего не найдено.


### 9.4. Главные выводы (*Cheat Sheet*)

* **Data Descriptor > `__dict__`**: Если у свойства есть `@property` или реализован `__set__`, ты не сможешь перебить его, просто записав значение в `__dict__`.
* **`__dict__` > Non-Data**: Обычные методы класса — это Non-data дескрипторы. Если ты создашь в экземпляре атрибут с тем же именем, что и метод, атрибут «спрячет» метод.
* **`__slots__` против `__dict__`**: Если используешь `__slots__`, Python не создает `__dict__` для объекта (если не указать его явно). Это экономит ~40-50% оперативной памяти при миллионах объектов.

## **10**. Метаклассы

**Метаклассы в Python —  это высший пилотаж Python, «магия» уровня Middle+/Senior, которую часто называют «классами для классов». Если коротко: обычный класс создает объекты, а метакласс создает классы.**

В Python всё является объектом, включая сами классы. Раз класс — это объект, у него должен быть «чертеж». Этот чертеж и есть **метакласс**.

* По умолчанию метаклассом для всех объектов является встроенный класс `type`.
* Схема: `Метакласс` $\rightarrow$ `Класс` $\rightarrow$ `Экземпляр`.

---

**Что вообще такое метакласс?**

Метакласс — это класс, экземплярами которого являются другие классы. Если обычный класс описывает поведение объектов, то метакласс описывает поведение классов.

Проще говоря:
- Объект создаётся классом.
- Класс создаётся метаклассом.

👉 Класс MyClass — это объект типа `type`. То есть `type` — это метакласс по умолчанию.

```
class MyClass:
    pass

print(type(MyClass))  # <class 'type'>
```

---

### 10.1. Как создать свой метакласс

Метакласс должен наследоваться от `type`. Обычно переопределяют метод `__new__`, который отвечает за создание объекта-класса.

#### Шаблон кода:

```
class MyMeta(type):
    def __new__(mcs, name, bases, attrs):
        # mcs — сам метакласс
        # name — имя создаваемого класса (строка)
        # bases — кортеж родительских классов
        # attrs — словарь атрибутов и методов класса

        # Пример: автоматически добавляем атрибут всем классам
        attrs['created_by_meta'] = True

        return super().__new__(mcs, name, bases, attrs)

# Применяем метакласс
class MyClass(metaclass=MyMeta):
    pass

print(MyClass.created_by_meta)  # True
```

### 10.2. Основные методы (Жизненный цикл)

При определении метакласса важны два метода:

1. **`__new__(mcs, name, bases, attrs)`**:<br/>
Вызывается **до** создания класса. Используется, чтобы изменить структуру класса (добавить методы, проверить наличие атрибутов).

2. **`__call__(cls, *args, **kwargs)`**:<br/>
Вызывается, когда вы создаете **экземпляр** уже созданного класса (например, `obj = MyClass()`). Часто используется для реализации паттерна **[Singleton](https://ru.wikipedia.org/wiki/%D0%9E%D0%B4%D0%B8%D0%BD%D0%BE%D1%87%D0%BA%D0%B0_(%D1%88%D0%B0%D0%B1%D0%BB%D0%BE%D0%BD_%D0%BF%D1%80%D0%BE%D0%B5%D0%BA%D1%82%D0%B8%D1%80%D0%BE%D0%B2%D0%B0%D0%BD%D0%B8%D1%8F)#Python)**.


### 10.3. Зачем это нужно на практике?

В обычном коде метаклассы нужны редко (как говорит Гвидо ван Россум: *"Если вы сомневаетесь, нужны ли они вам — они вам не нужны"*). Но они незаменимы в следующих случаях:

* **API и Фреймворки:** (например, Django ORM или SQLAlchemy). Когда вы пишете `models.CharField()`, метакласс «под капотом» превращает это в описание колонки базы данных.
* **Валидация:** Проверка, что подклассы реализуют определенные методы (хотя сейчас для этого чаще используют `abc.ABC`).
* **Автоматическая регистрация:** Добавление класса в глобальный реестр сразу при его создании.


### 10.4. Метаклассы vs Декораторы классов

| Характеристика | Декораторы классов | Метаклассы |
| --- | --- | --- |
| **Сложность** | Проще читать и писать | Сложнее, глубокая интеграция |
| **Наследование** | Не наследуются подклассами | **Наследуются** всеми потомками |
| **Цель** | Обертка / модификация готового класса | Определение правил «рождения» класса |

## **11**. Динамическое управление аттрибутами

В Python динамическое управление атрибутами — это мощный инструмент, который делает язык гибким. Вот краткая шпаргалка по основным способам.

**Что стоит помнить:**

1. Динамика замедляет код (поиск в словарях дороже прямых ссылок).
2. Чрезмерное использование `setattr` делает код сложным для отладки и "невидимым" для статических анализаторов (PyCharm может ругаться на несуществующие атрибуты).

### 11.1. Встроенные функции (Стандартный путь)

Самый безопасный и читаемый способ, особенно если имя атрибута хранится в переменной.

* **`setattr(obj, name, value)`**: Устанавливает значение.
* **`getattr(obj, name[, default])`**: Получает значение.
* **`hasattr(obj, name)`**: Проверяет существование.
* **`delattr(obj, name)`**: Удаляет атрибут.

```
class User: pass
u = User()

attr_name = "role"
setattr(u, attr_name, "admin")  # Эквивалент u.role = "admin"
print(u.role) # "admin"
```

---

### 11.2. Прямая работа с `__dict__`

Все атрибуты объекта (за редким исключением) хранятся в словаре `__dict__`.

* **Как использовать:** `obj.__dict__['key'] = value`
* **Особенность:** Это "грязный" способ. Он обходит механизмы защиты (например, `__setattr__`) и работает быстрее, но менее безопасно.

### 11.3. Магические методы (Перехват)

Если вы хотите контролировать процесс присвоения на уровне класса.

| Метод | Когда вызывается |
| --- | --- |
| `__setattr__(self, name, value)` | При **любой** попытке установить атрибут. |
| `__getattr__(self, name)` | Только если атрибут **не найден** обычным путем. |
| `__getattribute__(self, name)` | При **любом** обращении к атрибуту (опасно, легко уйти в рекурсию). |

> **Важно:** Внутри `__setattr__` всегда используйте `super().__setattr__` или правьте `self.__dict__` напрямую, чтобы избежать бесконечного цикла.

### 11.5. Динамическое создание методов

Метод — это просто функция, первым аргументом которой является `self`.

```
import types

def say_hi(self):
    return f"Hi, I'm {self.name}"

u = User()
u.name = "Alice"
# Привязываем функцию к конкретному экземпляру
u.greet = types.MethodType(say_hi, u)

print(u.greet()) # "Hi, I'm Alice"
```
